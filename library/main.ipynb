## METODE SECANT

```python
from secant_library import SecantMethod
import math
```

## Contoh Soal

```python
#Masukkan Fungsi
def f(x):
    return x**3 - x - 2

#Jalankan Metode Secant
secant = SecantMethod (
    func=f,
    x0=1,        # tebakan awal pertama
    x1=2,        # tebakan awal kedua
    tol=1e-6,    # toleransi error
    max_iter=20  # maksimum iterasi
)
secant.iterate()
secant.show_solution()
secant.plot()
```
